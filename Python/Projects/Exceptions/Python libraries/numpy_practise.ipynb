Concatenation (combine array)

```python
b1=np.array([56,78,89])
b2=np.array([65,76,98])

comb=np.concatenate((b1,b2))
comb
```

```python
print(np.split(comb,2)) #to seperate the data into groups use split.

for arr in np.split(comb,3): # 'array split does always result in an equal division only'.
    print(arr)
```

```python
b1=np.array([[56,78,89],[34,45,67]])  # using axis

b2=np.array([[85,78,90],[89,98,78]])

comb1=np.concatenate((b1,b2),axis=1)
print(comb1)
# #output= [[56 78 89 85 78 90]
#  [34 45 67 89 98 78]]

comb0=np.concatenate((b1,b2),axis=0)
print(comb0)

# #output=[[56 78 89]
#  [34 45 67]
#  [85 78 90]
#  [89 98 78]]
```
